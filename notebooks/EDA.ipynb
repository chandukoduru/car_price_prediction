# CarDekho Dataset: Exploratory Data Analysis

Use this notebook to inspect the licensed `dataset/car_data.csv` before training. It intentionally does not generate sample or dummy data. The training script performs the full validated cleaning, charts, model comparison, and serialization workflow.

```python
from pathlib import Path
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

DATA_PATH = Path('../dataset/car_data.csv')
required = {'Car_Name', 'Year', 'Selling_Price', 'Present_Price', 'Driven_kms', 'Fuel_Type', 'Selling_type', 'Transmission', 'Owner'}
if not DATA_PATH.is_file():
    raise FileNotFoundError('Add the licensed CarDekho CSV at dataset/car_data.csv first.')
data = pd.read_csv(DATA_PATH)
missing = required.difference(data.columns)
if missing:
    raise ValueError(f'Missing required dataset columns: {sorted(missing)}')
data.head()
```

```python
data.info()
display(data.describe(include='all').T)
display(data.isna().sum().rename('missing values').to_frame())
print(f'Duplicate rows: {data.duplicated().sum()}')
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
sns.histplot(data=data, x='Selling_Price', kde=True, ax=axes[0, 0])
sns.boxplot(data=data, x='Fuel_Type', y='Selling_Price', ax=axes[0, 1])
sns.boxplot(data=data, x='Transmission', y='Selling_Price', ax=axes[1, 0])
sns.scatterplot(data=data, x='Year', y='Selling_Price', alpha=.65, ax=axes[1, 1])
axes[0, 0].set_title('Selling price distribution')
axes[0, 1].set_title('Fuel type vs selling price')
axes[1, 0].set_title('Transmission vs selling price')
axes[1, 1].set_title('Manufacturing year vs selling price')
plt.tight_layout()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=data, x='Driven_kms', y='Selling_Price', alpha=.65)
plt.title('Kilometers driven vs selling price')
plt.ylabel('Selling price (check dataset units)')
plt.tight_layout()
```

Before training, confirm the dataset source, license, currency/units, and category spelling. Do not treat exploratory relationships as causal or as guarantees of resale value.
